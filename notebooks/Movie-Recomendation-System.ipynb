```python
import ast
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies.head()
```

```python
movies.shape
```

```python
credits.shape
```

```python
data = movies.merge(credits,on='title')
```

```python
data.shape
```

```python
data2 = data[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
data2.isna().sum()
```

```python
data2.dropna(inplace=True)
```

```python
data2.duplicated().sum()
```

```python
data2.genres[0]
```

```python
def convert(obj):
    lst = []
    for i in ast.literal_eval(obj):
        lst.append(i['name'])
    return lst
```

```python
data2['genres'] = data2.genres.apply(convert)
```

```python
data2['keywords'] = data2.keywords.apply(convert)
```

```python
def convert3(obj):
    lst = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter!=3:
            lst.append(i['name'])
            counter+=1
        else:
            break
    
    return lst
```

```python
data2['cast'] = data2['cast'].apply(convert3)
```

```python
def convert4(obj):
    lst = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            lst.append(i['name'])
            break
    return lst
```

```python
data2['crew'] = data2['crew'].apply(convert4)
```

```python
data2
```

```python
data2['overview'][0]
```

```python
data2['genres'] = data2['genres'].apply(lambda x:[i.replace(" ","") for i in x])
data2['keywords'] = data2['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
data2['cast'] = data2['cast'].apply(lambda x:[i.replace(" ","") for i in x])
data2['crew'] = data2['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
data2
```

```python
data2['overview'] = data2['overview'].apply(lambda x: x.split())
```

```python
data2['tags'] = data2['overview'] + data2['genres'] + data2['keywords'] + data2['cast'] + data2['crew']
```

```python
data2
```

```python
new_data = data2[['movie_id','title','tags']]
```

```python
new_data['tags'] = new_data['tags'].apply(lambda x:" ".join(x))
```

```python
new_data['tags'] = new_data['tags'].apply(lambda x:x.lower())
```

```python
new_data['tags']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_data['tags']).toarray()
```

```python
vectors[0]
```

```python
len(cv.get_feature_names_out())
```

```python
print(list(cv.get_feature_names_out()))
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
ps.stem('')
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_data['tags'] = new_data['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
simalarity = cosine_similarity(vectors)
```

```python
simalarity[0]
```

```python
def recommend(movie):
    movie_index = new_data[new_data['title'] == movie].index[0]
    distance = simalarity[movie_index]
    movie_list = sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movie_list:
        print(new_data.iloc[i[0]].title)
    
```
